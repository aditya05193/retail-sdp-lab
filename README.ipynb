**These two lines add information about the source file to every Bronze record.**

F.col("_metadata.file_path").alias("_source_file")

_metadata.file_path is automatically provided by Databricks when reading files.

It stores the full path of the file that contained the record, for example:

/Volumes/workspace/retail_dev/source_files/orders/batch_003/orders_003.json

We rename it to _source_file.

**Why it helps:** if an order is invalid, you can identify exactly which file delivered it.

F.col("_metadata.file_modification_time").alias(
    "_source_file_modified_at"
)

_metadata.file_modification_time is also provided by Databricks. It records when the source file was last modified, for example:
2026-11-14 09:20:30

We rename it to _source_file_modified_at.

**Why it helps:** it shows when the source data producer created or updated the file. This supports investigation, freshness monitoring, and backfill analysis.

For each Bronze row, you will have both business data and file audit information:

order_id	quantity	_source_file	_source_file_modified_at

O0001	2	.../orders_001.json	2026-11-14 09:20:30


**The underscore prefix is a naming convention. It means these are technical metadata columns, not business columns supplied by the retail source system.**
